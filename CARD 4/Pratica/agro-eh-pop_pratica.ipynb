```python
def categorizar_producao(valor):
    if pd.isna(valor):
        return "Sem dado"
    elif valor < 100:
        return "Baixa produção"
    elif valor < 1000:
        return "Média produção"
    else:
        return "Alta produção"
```

```python
agro_long["Categoria_Producao"] = agro_long["Production"].apply(categorizar_producao) # classificando as produções
```
