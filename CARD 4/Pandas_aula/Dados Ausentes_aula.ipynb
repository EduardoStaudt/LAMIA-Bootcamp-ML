```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.DataFrame({'A':[1, 2, np.nan], 'B': [5, np.nan, np.nan], 'C':[1,2,3]})
```

```python
df
```

```python
df.dropna() # padrão é axis=0 portanto se tem nan na linha ela é excluida
```

```python
df.dropna(axis=1) # assim se tiver nan na coluna ela será excluida
```

```python
df.dropna(thresh=2) # vai excluir as linhas que possuirem a quantidae pasada de valores faltantes aqui foi 2 mas poderia ser qualquer um
```

```python
df['A'].fillna(value=df['A'].mean())
```

```python
df .fillna(method='ffill') # foward fill substitui o nan pelo anterior a ele na coluna
```

```python
# assim apos eu rodar aparece esse aviso de que o method é não esta mais sendo muito utilizado e em versões futuras ele pode deixar de funcionar no pandas
df.ffill() 3 essa é a versão atualaizada
```
