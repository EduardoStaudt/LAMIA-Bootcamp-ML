# Standard Deviation and Variance

```python
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 50.0, 10000) # gera com dist normal

plt.hist(incomes, 50)
plt.show()
```

```python
incomes.std() # calc desvio padrao
```

```python
incomes.var() # calc variancia
```
