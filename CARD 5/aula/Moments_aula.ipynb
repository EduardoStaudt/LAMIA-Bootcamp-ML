# Moments: Mean, Variance, Skew, Kurtosis

Create a roughly normal-distributed random set of data:

```python
# principais momentos
#1- média
#2- variância
#3- skew = o quão distoante a distyribuição estaá elea se enclinando pra algum lado? positivo ou negativo
#4- kurtosis = mede os picos de uam distraibuição e isso mostra como esses picos influenciam na média
```

```python
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000)

plt.hist(vals, 50)
plt.show()
```

The first moment is the mean; this data should average out to about 0:

```python
np.mean(vals) # media
```

The second moment is the variance:

```python
np.var(vals) # vaiancia
```

The third moment is skew - since our data is nicely centered around 0, it should be almost 0:

```python
import scipy.stats as sp # precisa do scipy
sp.skew(vals) # distorção da média
```

The fourth moment is "kurtosis", which describes the shape of the tail. For a normal distribution, this is 0:

```python
sp.kurtosis(vals) # influencia dospicos na media
```
