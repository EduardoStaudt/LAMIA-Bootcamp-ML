# Examples of Data Distributions

## Uniform Distribution

```python
# Distribuição uniforme praticamente todos com a mesma chance
import numpy as np
import matplotlib.pyplot as plt

values = np.random.uniform(-10.0, 10.0, 100000)
plt.hist(values, 50)
plt.show()
```

## Normal / Gaussian

Visualize the probability density function:

```python
# distribuiicao normal valores maior dominio na média
from scipy.stats import norm
import matplotlib.pyplot as plt

x = np.arange(-3, 3, 0.001) # com arange
plt.plot(x, norm.pdf(x))
plt.show()
```

Generate some random numbers with a normal distribution. "mu" is the desired mean, "sigma" is the standard deviation:

```python
# Histograma
import numpy as np
import matplotlib.pyplot as plt

mu = 5.0
sigma = 2.0
values = np.random.normal(mu, sigma, 10000) # com random normal escolhen do aonde centarr a média
plt.hist(values, 50)
plt.show()
```

## Exponential PDF / "Power Law"

```python

from scipy.stats import expon
import matplotlib.pyplot as plt

x = np.arange(0, 10, 0.001)
plt.plot(x, expon.pdf(x))
plt.show()
```

## Binomial Probability Mass Function

```python
# para dados discretos a binomial trabalho com sucesso ou falha, bom pra boolean
from scipy.stats import binom
import matplotlib.pyplot as plt

n, p = 10, 0.5
x = np.arange(0, 10, 0.001)
plt.plot(x, binom.pmf(x, n, p))
plt.show()
```

## Poisson Probability Mass Function

Example: My website gets on average 500 visits per day. What's the odds of getting 550?

```python
# l3gal pra procuar chances no grafico pois ele permite pesquisar coisas que vc ainda não sabe usa-se para calcular quantidade de eventos em um intervalo 
from scipy.stats import poisson
import matplotlib.pyplot as plt

mu = 500
x = np.arange(400, 600, 0.5)
plt.plot(x, poisson.pmf(x, mu))
plt.show()
```
