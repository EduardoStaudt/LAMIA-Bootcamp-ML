```python
import numpy as np

incomes = np.random.normal(27000, 15000, 10000) # gere dados com dist normal nessas faixas
np.mean(incomes) # tira a média
```

```python
import matplotlib.pyplot as plt # lib de graficos
plt.hist(incomes, 50) # separando em 50 grupos tranformando o nums continuos em discretos e centra na medis
plt.show() # exibe
```

```python
np.median(incomes)
```

```python
incomes = np.append(incomes, [1000000000])
```

```python
np.median(incomes) # mediana tende a ser mais correta para saber centros reais
```

```python
np.mean(incomes) #a média geralmente mascarta a verdade
```
