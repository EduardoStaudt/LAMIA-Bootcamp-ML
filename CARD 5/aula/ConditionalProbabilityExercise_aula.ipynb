# Conditional Probability Activity & Exercise

```python
from numpy import random
random.seed(0) # so pra manter os mesmos aleatorios

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0} #ages
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0} # qtd  compras
totalPurchases = 0 #soma tudo
for _ in range(100000): # atribui de forma aleatoria uma decada e um valor
    ageDecade = random.choice([20, 30, 40, 50, 60, 70]) 
    purchaseProbability = float(ageDecade) / 100.0 # chance de comprar idade divido por 100
    totals[ageDecade] += 1
    if (random.random() < purchaseProbability): #aleatório tem de ser menor do que a chance de comprar
        totalPurchases += 1
        purchases[ageDecade] += 1
```

```python
PEF = float(purchases[30]) / float(totals[30]) # vendas com 30 / total de pessoas
print('P(purchase | 30s): ' + str(PEF))
```

```python
PF = float(totals[30]) / 100000.0 # total de pessoas idade pelo total geral
print("P(30's): " +  str(PF))
```

```python
PE = float(totalPurchases) / 100000.0 # total de vendas pelo total de pessoas
print("P(Purchase):" + str(PE))
```

```python
print("P(30's, Purchase)" + str(float(purchases[30]) / 100000.0)) # total de vendas 30 pelo total de pessoas
```

```python
print("P(30's)P(Purchase)" + str(PE * PF)) # probabilidade de comprar algo estando nos 30
```

```python
print((purchases[30] / 100000.0) / PF)
```
