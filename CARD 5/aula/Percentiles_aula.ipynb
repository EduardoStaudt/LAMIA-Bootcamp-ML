# Percentiles

```python
import numpy as np
import matplotlib.pyplot as plt
#  o percentil permite indentifacr a disposição dos dados 
vals = np.random.normal(0, 0.5, 10000)

plt.hist(vals, 50)
plt.show()
```

```python
np.percentile(vals, 50) # testando o percentil 50 que é a mediana basicamenet
```

```python
np.percentile(vals, 90) # testando o percentil 90% menor que 0.64
```

```python
np.percentile(vals, 20) # testando o percentil 20% menor que -0.40
```
