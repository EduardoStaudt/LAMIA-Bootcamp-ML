# Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("http://media.sundog-soft.com/SelfDriving/FuelEfficiency.csv")
gear_counts = df['# Gears'].value_counts()

gear_counts.plot(kind='bar')
plt.show()
```

```python
import seaborn as sns # importo
sns.set()
```

```python
gear_counts.plot(kind='bar') # plot(kind=tipo)
```

```python
df.head()
```

```python
sns.displot(df['CombMPG']) # graf de rendimento
```

```python
df2 = df[['Cylinders', 'CityMPG', 'HwyMPG', 'CombMPG']] # passo só as colunas de interresse
df2.head()
```

```python
sns.pairplot(df2, hue='Cylinders', height=2.5);
```

```python
sns.scatterplot(x="Eng Displ", y = "CombMPG", data=df) # correlation
```

```python
sns.jointplot(x="Eng Displ", y="CombMPG", data=df) # com sobreposição dos hist
```

```python
sns.lmplot(x="Eng Displ", y = "CombMPG", data=df) # dispersão mais regressão linear
```

```python
sns.set(rc={'figure.figsize': (15, 5)})
ax = sns.boxplot(x='Mfr Name', y='CombMPG', data=df) # boxpot de combustivel por marca
plt.xticks(rotation=45)
```

```python
ax = sns.swarmplot(x='Mfr Name', y='CombMPG', data=df)
ax.tick_params(axis='x', rotation=45) # como dist evidente
```

```python
ax = sns.countplot(x='Mfr Name', data=df) # modelos de caros
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
```

```python
df2 = df.pivot_table(index='Cylinders', columns='Eng Displ', values='CombMPG', aggfunc='mean')
sns.heatmap(df2) # heatmap de cilindros por cilidrada 
```
