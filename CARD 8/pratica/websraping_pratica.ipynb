```python
# imports
import requests
from bs4 import BeautifulSoup as bs
```

```python
r = requests.get("https://www.lamia-edu.com/")
lamia = bs(r.content)
print(lamia.prettify())
```

```python
url = "https://www.lamia-edu.com/" # link

pagina = requests.get(url) 
soup = bs(pagina.text, "html.parser")

foto = soup.find("img", alt="Foto do Thiago Naves") # Encontra a tag da foto
url_foto = url + foto["src"].lstrip("/")# Aqui removi a barra de inicio e juntei com o endereço do site do LAMIA

imagem = requests.get(url_foto) # gerei a requisição

with open("lenda_thiago_naves.jpg", "wb") as arquivo: # e escrevi com o with o jpg
    arquivo.write(imagem.content)
```

```python
pagina = requests.get(url)
soup = bs(pagina.text, "html.parser")

orientadores = soup.select("h3.text-white.text-xl.font-bold.leading-tight.drop-shadow-sm") # achei pelo inpecionar no dev tools

nomes = [] # lita pra guardar os nomes
for orientador in orientadores:
    nomes.append(orientador.get_text(strip=True))# Vai guardando os nomes 

import pandas as pd

df = pd.DataFrame(nomes, columns=["Nome"])# Monta o df
df.head(10) # pra mostra
```

```python
#sem ver pelo inpecionar
redes_sociais = soup.select("""a[href*="facebook.com"], a[href*="instagram.com"], a[href*="github.com"], a[href*="linkedin.com"]""")

for rede in redes_sociais:
    print(rede["href"]) # aqui aparece uns repetido que tem pelo site

print("---------")

#vendo pelo inspecionar
redes_sociais = soup.select('a[class~="hover:scale-150"].duration-300.scale-125') 

for rede in redes_sociais:
    print(rede["href"]) #Printando
```
