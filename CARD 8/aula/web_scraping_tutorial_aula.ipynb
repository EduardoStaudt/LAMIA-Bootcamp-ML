## Load the webpage

```python
r = requests.get('https://keithgalli.github.io/web-scraping/webpage.html')
webpage = bs(r.content)
print(webpage.prettify())# carreguei e organizei o html
```

## Grab all of the social links from the webpage
Do this in at least 3 different ways

```python
# com select no ul
links = webpage.select('ul.socials a')
actual_links = [link['href'] for link in links] # limpa pega só o endereço mesmo
actual_links
```

```python
# com find
ulist = webpage.find("ul", attrs={"class": "socials"}) # mesma coisa do de cima mas sem o selector de CSS
links = ulist.find_all("a") #pega os links
actual_links = [link['href'] for link in links] # limpa
actual_links
ulist
```

```python
# select no li
links = webpage.select("li.social a")
actual_links = [link['href'] for link in links]
actual_links
links
```

## Scrape the Table

```python
import pandas as pd

table = webpage.select('table.hockey-stats')[0] # pega a primeira tb da pagina
columns = table.find('thead').find_all('th') # quarda as colunas
columns_names = [c.string for c in columns] # pega os nomes

table_rows = table.find('tbody').find_all('tr') # pega as linha

l = []
# mmonta a tb
for tr in table_rows:
    td = tr.find_all('td')
    row = [c.get_text().strip() for c in td]
    l.append(row)
# e salva
df = pd.DataFrame(l, columns=columns_names)
df.loc[df['Team'] != 'Did not play']
df.head()
```

## Grab all fun facts that use word 'is'

```python
facts = webpage.select("ul.fun-facts li") # ir até a lista de lisks
facts_with_is = [fact.find(string=re.compile("is")) for fact in facts]# o compile procuar os is
facts_with_is = [fact.find_parent().get_text() for fact in facts_with_is if fact]
facts_with_is
```

### Solve the mistery challenge

```python
files = webpage.select("div.block a")
relative_files = [file['href'] for file in files] # extrai os links

url = "https://keithgalli.github.io/web-scraping/"
for f in relative_files:
    full_url = url + f # Concatena o a url do portfolio + um dos links
    page = requests.get(full_url)
    bs_page = bs(page.content)
    secret_word_element = bs_page.find("p", attrs={"id": "secret-word"}) #localiza o paragrafo com a secret-word
    secret_word = secret_word_element.string # coleta em string
    print(secret_word)
```
